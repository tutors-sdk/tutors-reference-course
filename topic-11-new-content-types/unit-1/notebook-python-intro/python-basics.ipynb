## Variables and Data Types

Python supports several built-in data types. Let's explore the most common ones.

```python
# Basic variable assignment
name = "Tutors Platform"
version = 15.0
is_active = True
features = ["labs", "talks", "notes", "notebooks"]

print(f"Name: {name}")
print(f"Version: {version}")
print(f"Active: {is_active}")
print(f"Features: {', '.join(features)}")
```

## Working with Lists

Lists are ordered, mutable collections. They're one of the most versatile data structures in Python.

```python
# List operations
scores = [85, 92, 78, 95, 88]

print(f"Scores: {scores}")
print(f"Average: {sum(scores) / len(scores)}")
print(f"Highest: {max(scores)}")
print(f"Lowest: {min(scores)}")
print(f"Sorted: {sorted(scores)}")
```

## Functions

Functions let you organize and reuse code. Python uses the `def` keyword.

```python
def calculate_grade(scores):
    """Calculate letter grade from a list of scores."""
    average = sum(scores) / len(scores)
    if average >= 90:
        return "A", average
    elif average >= 80:
        return "B", average
    elif average >= 70:
        return "C", average
    elif average >= 60:
        return "D", average
    else:
        return "F", average

students = {
    "Alice": [85, 78, 92, 75],
    "Bob": [95, 88, 91, 90],
    "Charlie": [72, 68, 80, 76]
}

for name, scores in students.items():
    grade, avg = calculate_grade(scores)
    print(f"{name}: {grade} ({avg})")
```

## List Comprehensions

A concise way to create lists based on existing sequences.

```python
# List comprehension examples
squares = [x**2 for x in range(1, 11)]
print(f"Squares: {squares}")

even_squares = [x**2 for x in range(1, 11) if x % 2 == 0]
print(f"Even squares: {even_squares}")

# Dictionary comprehension
words = ["hello", "world", "python", "tutors"]
word_lengths = {w: len(w) for w in words}
print(f"Word lengths: {word_lengths}")
```

## Error Handling

Python uses `try`/`except` blocks to handle errors gracefully.

```python
def safe_divide(a, b):
    """Safely divide two numbers with error handling."""
    try:
        result = float(a) / float(b)
        return f"{result:.2f}"
    except ZeroDivisionError:
        return "Cannot divide by zero!"
    except ValueError as e:
        return f"Invalid input: {e}"

print(f"10 / 3 = {safe_divide(10, 3)}")
print(f"10 / 0 = {safe_divide(10, 0)}")
print(f"10 / a = {safe_divide(10, 'a')}")
```

## Mathematical Notation

Notebooks support LaTeX for mathematical expressions:

The quadratic formula: $x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}$

Let's implement it:

```python
import math

def quadratic(a, b, c):
    """Solve ax² + bx + c = 0 using the quadratic formula."""
    discriminant = b**2 - 4*a*c
    if discriminant < 0:
        return None, None
    x1 = (-b + math.sqrt(discriminant)) / (2*a)
    x2 = (-b - math.sqrt(discriminant)) / (2*a)
    return x1, x2

a, b, c = 1, -5, 6
x1, x2 = quadratic(a, b, c)
print(f"Solving: {a}x\u00b2 + {b}x + {c} = 0")
print(f"Solutions: x = {x1} and x = {x2}")
```
